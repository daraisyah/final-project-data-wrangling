```python
# Buat data dictionary
data_dict = []

for kolom in df.columns:
    info = {}
    info["variable"]    = kolom
    info["data_type"]   = str(df[kolom].dtype)
    info["n_missing"]   = df[kolom].isna().sum()
    info["n_unique"]    = df[kolom].nunique()

    # Kalau kolom angka, tambahkan min, max, mean
    if df[kolom].dtype in ["int64", "float64"]:
        info["min"]  = df[kolom].min()
        info["max"]  = df[kolom].max()
        info["mean"] = round(df[kolom].mean(), 3)
    else:
        info["min"]  = "-"
        info["max"]  = "-"
        info["mean"] = "-"

    # Contoh nilai
    info["contoh_nilai"] = str(df[kolom].dropna().unique()[:3].tolist())

    data_dict.append(info)

dd = pd.DataFrame(data_dict)

# Tambahkan deskripsi manual
deskripsi = {
    "track_id"                : "ID unik lagu di Spotify",
    "track_name"              : "Judul lagu",
    "track_artist"            : "Nama artis",
    "track_popularity"        : "Skor popularitas lagu (0-100)",
    "track_album_id"          : "ID album di Spotify",
    "track_album_name"        : "Nama album",
    "track_album_release_date": "Tanggal rilis album",
    "playlist_name"           : "Nama playlist",
    "playlist_id"             : "ID playlist di Spotify",
    "playlist_genre"          : "Genre playlist",
    "playlist_subgenre"       : "Subgenre playlist",
    "danceability"            : "Seberapa cocok lagu untuk menari (0-1)",
    "energy"                  : "Intensitas dan aktivitas lagu (0-1)",
    "key"                     : "Nada dasar lagu (0-11)",
    "loudness"                : "Kekerasan suara dalam dB",
    "mode"                    : "Mayor=1 atau Minor=0",
    "speechiness"             : "Proporsi kata-kata (0-1)",
    "acousticness"            : "Seberapa akustik lagu tersebut (0-1)",
    "instrumentalness"        : "Kemungkinan lagu tanpa vokal (0-1)",
    "liveness"                : "Kemungkinan direkam secara live (0-1)",
    "valence"                 : "Positivitas lagu (0-1)",
    "tempo"                   : "Kecepatan lagu dalam BPM",
    "duration_ms"             : "Durasi lagu dalam milidetik",
}

nilai_harapan = {
    "track_id"                : "String unik dan tidak kosong",
    "track_name"              : "String",
    "track_artist"            : "String",
    "track_popularity"        : "Integer 0 sampai 100",
    "track_album_id"          : "String",
    "track_album_name"        : "String",
    "track_album_release_date": "YYYY atau YYYY-MM atau YYYY-MM-DD",
    "playlist_name"           : "String",
    "playlist_id"             : "String",
    "playlist_genre"          : "pop, rap, rock, latin, r&b, edm",
    "playlist_subgenre"       : "Kategori teks",
    "danceability"            : "Float 0.0 sampai 1.0",
    "energy"                  : "Float 0.0 sampai 1.0",
    "key"                     : "Integer 0 sampai 11",
    "loudness"                : "Float",
    "mode"                    : "0 atau 1",
    "speechiness"             : "Float 0.0 sampai 1.0",
    "acousticness"            : "Float 0.0 sampai 1.0",
    "instrumentalness"        : "Float 0.0 sampai 1.0",
    "liveness"                : "Float 0.0 sampai 1.0",
    "valence"                 : "Float 0.0 sampai 1.0",
    "tempo"                   : "Float lebih dari 0",
    "duration_ms"             : "Integer lebih dari 0",
}

dd["deskripsi"]      = dd["variable"].map(deskripsi)
dd["nilai_harapan"]  = dd["variable"].map(nilai_harapan)

# Tampilkan
print("=== DATA DICTIONARY ===")
dd[["variable", "deskripsi", "data_type", "nilai_harapan",
    "n_missing", "n_unique", "min", "max", "mean"]]
```

```python
# Simpan ke CSV
dd.to_csv("data_dictionary.csv", index=False)
print("data_dictionary.csv berhasil disimpan")
```
